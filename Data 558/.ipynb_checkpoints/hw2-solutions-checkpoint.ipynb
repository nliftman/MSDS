## Exercise 3

- Assume that d=1 and n=1. The sample is then of size 1 and boils down to just (x,y). The function F writes simply as
$$
F(\beta) = \frac{1}{2}(y-x\beta)^2 + \lambda \beta^2
$$
Compute and write down the gradient $\nabla F$ of $F$.

__Solution:__
$$
\nabla F(\beta) = -x(y-x\beta)+2\lambda \beta1      
$$

- Assume now that d>1 and n>1. Using the previous result and the linearity of differentiation, compute and write own the gradient $\nabla F$ of $F$.

__Solution:__

$$\nabla F = -\frac{1}{n}X^T(Y-X\beta) + \frac{2}{n}\lambda \beta.$$

```python
from sklearn.model_selection import train_test_split
from sklearn import preprocessing
```

```python
penguins = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv',
            sep=',', header=0)
penguins = penguins.dropna()
```

```python
X = penguins.drop('body_mass_g', axis=1)
X = pd.get_dummies(X, drop_first=True)
y = penguins['body_mass_g']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```

```python
scaler = preprocessing.StandardScaler().fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)

scaler = preprocessing.StandardScaler().fit(y_train.values.reshape(-1, 1))
y_train = scaler.transform(y_train.values.reshape(-1, 1)).reshape((-1))
y_test = scaler.transform(y_test.values.reshape(-1, 1)).reshape((-1))
```

Implement gradient and GD algorithm with max. iteration stopping criterion.

```python
def computegrad(beta, lam, X=X_train, y=y_train):
    resid = y-np.dot(X, beta)
    n = len(X)
    return -1/n*X.T.dot(resid) + 2/n*lam*beta
```

```python
def graddescent(beta_init, eta, lam, max_iter=1000):
    """Run gradient descent with a fixed step size
    Inputs:
    - beta_init: Starting point
    - eta: Step size (a constant)
    - max_iter: Maximum number of iterations to perform
    Output:
    - beta_vals: Matrix of estimated betas at each iteration,
                with the most recent values in the last row.
    """
    beta = beta_init
    grad_beta = computegrad(beta, lam)
    beta_vals = [beta]
    iter_num = 0
    while iter_num < max_iter:
        beta = beta - eta*grad_beta
        beta_vals.append(beta)
        grad_beta = computegrad(beta, lam)
        iter_num += 1

    return np.array(beta_vals)
```

```python
# also implement objective function
def ridge_obj(beta, lam, x=X_train, y=y_train):
    return 1/(2*len(X))*sum((y-x.dot(beta))**2) + lam/len(X)*np.linalg.norm(beta)**2
```

Run with $\lambda = -5$.

```python
eta = 0.5
max_iter = 1000
lam = -5
d = X_train.shape[1]
beta_init = np.zeros(d)

iterates = graddescent(beta_init, eta, lam, max_iter=max_iter)
```

```python
obj_values = [ridge_obj(x, lam) for x in iterates] # objective values at every iterate

grad_norm = [np.linalg.norm(computegrad(x, lam)) for x in iterates] # gradient norm at every iterate
```

```python
plt.plot(obj_values) # doesn't converge!
plt.xlabel('Iteration number')
plt.ylabel(r'Ridge objective $F(\beta)$')

plt.show()
```

Run with $\lambda = 0.05$.

```python
eta = 0.5
max_iter = 1000
lam = 0.05
d = X_train.shape[1]
beta_init = np.zeros(d)

iterates = graddescent(beta_init, eta, lam, max_iter=max_iter)
```

```python
obj_values = [ridge_obj(x, lam) for x in iterates] # objective values at every iterate

grad_norm = [np.linalg.norm(computegrad(x, lam)) for x in iterates] # gradient norm at every iterate
```

```python
plt.plot(obj_values) # converges (quickly)
plt.xlabel('Iteration number')
plt.ylabel(r'Ridge objective $F(\beta)$')

plt.show()
```

```python
plt.plot(obj_values[:30]) # zoom in on previous plot
plt.xlabel('Iteration number')
plt.ylabel(r'Ridge objective $F(\beta)$')

plt.show()
```

```python
plt.xlabel('Iteration number') # gradient norm goes to zero
plt.ylabel(r'Gradient norm $\nabla F(\beta)$')
plt.plot(grad_norm)

plt.show()
```

Compare to sklearn's Ridge().

```python
from sklearn.linear_model import Ridge

n = len(y_train)
alpha = lam * 2 # scale alpha so sklearn objective matches ours

ridge = Ridge(alpha=alpha, fit_intercept=False)
ridge.fit(X_train, y_train)

# coefficients match
print('sklearn coeffs:\n {}'.format(np.around(ridge.coef_, 3)))
print('our coeffs:\n {}'.format(np.around(iterates[-1], 3)))
```

```python
# objective values match
print('objective at sklearn coefficients:\n {:.6f}'.format(ridge_obj(ridge.coef_, lam)))
print('objective at our coefficients:\n {:.6f}'.format(ridge_obj(iterates[-1], lam)))
```

Run over many values of $\eta$.

```python
eta_range = np.logspace(-4, 1, 21)

max_iter = 1000
lam = 5
d = X_train.shape[1]

final_obj_vals = []
for eta in eta_range:
    # run GD with this choice of eta
    beta_init = np.zeros(d)
    iterates = graddescent(beta_init, eta, lam, max_iter=max_iter)

    # check final
    final_obj = ridge_obj(iterates[-1], lam)
    final_obj_vals.append(final_obj)
```

```python
'''
What's going on here:

- Pick the step size too small and the alg doesn't converge within 1000 iterations.
- Pick it too large and the objective values diverge, hence the NaNs.

'''

final_obj_vals
```

```python
na_ix = np.isnan(final_obj_vals)
```

```python
finite_vals = np.array(final_obj_vals)[~na_ix]
good_etas = eta_range[~na_ix]

best_ix = np.argmin(finite_vals)
best_eta = good_etas[best_ix]

print(best_eta) # this will vary, depending on the range you searched over
```

```python
# compare to sklearn
iterates = graddescent(beta_init, best_eta, lam, max_iter)
```

```python
# coeffs
print('sklearn coeffs:\n {}'.format(np.around(ridge.coef_, 3)))
print('our coeffs:\n {}'.format(np.around(iterates[-1], 3)))
```

Modify implementation to have stopping criterion in terms of gradient norm.

```python
# grad descent with gradient norm stopping criterion
def graddescent(beta_init, eta, lam, tol=0.005):
    """Run gradient descent with a fixed step size
    Inputs:
    - beta_init: Starting point
    - eta: Step size (a constant)
    - max_iter: Maximum number of iterations to perform
    Output:
    - beta_vals: Matrix of estimated betas at each iteration,
                with the most recent values in the last row.
    """
    beta = beta_init
    grad_beta = computegrad(beta, lam)
    beta_vals = [beta]
    iter_num = 0
    while np.linalg.norm(grad_beta) > tol:
        beta = beta - eta*grad_beta
        beta_vals.append(beta)
        grad_beta = computegrad(beta, lam)
        iter_num += 1

    return np.array(beta_vals)
```

```python
eta = 0.5
tol = 0.005
lam = 0.05
d = X_train.shape[1]
beta_init = np.zeros(d)
iterates = graddescent(beta_init, eta, lam, tol=tol)
```

```python
obj_trace = [ridge_obj(b, lam) for b in iterates]
```

```python
'''
In general, the results after modifying the implementation
should look very much the same as above.

The main difference is that the iteration number is no
longer fixed. Rather, we stop when we are sufficiently close
to a critical point. In this example, this means the algorithm
will terminate much faster than when we had set a large value
of max_iter.

'''

plt.plot(obj_trace) # terminates after ~50 iterations, not 1000 as above
plt.ylabel('Ridge objective')
plt.xlabel('GD iteration')
plt.show()
```

```python
# check modified implementation against sklearn

'''
There will likely be some noticeable error here.

This is because we stop when the gradient norm is not exactly zero,
and in particular the tolerance we set is looser than what sklearn
uses. To check that the implementation still matches sklearn's result,
one could decrease the error tolerance above, say by a factor of 100.
'''
from sklearn.linear_model import Ridge

n = len(y_train)
alpha = lam * 2
ridge = Ridge(alpha=alpha, fit_intercept=False)

ridge.fit(X_train, y_train)

print('sklearn coeffs:\n {}'.format(np.around(ridge.coef_, 3)))
print('our coeffs:\n {}'.format(np.around(iterates[-1], 3)))
```

```python
import statsmodels
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
auto = pd.read_csv('https://www.statlearning.com/s/Auto.csv', na_values='?')
```

```python
auto.shape
```

```python
formula = 'mpg ~ weight'
```

```python
lm = smf.ols(formula=formula, data=auto)
```

```python
res = lm.fit()
```

```python
res.summary()
```

From the above summary we conclude that there is a statistically significant, negative association between weight and mpg.

```python
statsmodels.graphics.regressionplots.plot_fit(res, exog_idx='weight');
```

```python
fitted = res.fittedvalues
resids = res.resid

plt.plot(fitted, resids, '.')
plt.ylabel('Residual')
plt.xlabel('Fitted value');
```

We observe both a trend and heteroskedasticity in the residuals as a function of the fitted value. This suggests that the true relationship between these variables may not be linear, and that the OLS assumption of constant variance is likely violated.
