#### (a) Assume that d = 1 and n = 1. The sample is then of size 1 and boils down to just (x, y). The function F writes simply as F (β) = log(1 + exp(−yx β) + λβ2 . Compute and write down the gradient ∇F of F 

The gradient of F(B) is: $\nabla F(\beta) = {-yxe^{-yx\beta}}/{1+ e^{-yx\beta}} + 2\lambda(\beta)$. We can simplify this to: $\nabla F(\beta) = (-yx)/(1+ e^{yx\beta}) + 2\lambda(\beta)$.

#### (b) Assume now that d > 1 and n > 1. Using the previous result and the linearity of differentiation, compute and write down the gradient ∇F (β) of F .

```python
from sklearn.model_selection import train_test_split
from sklearn import preprocessing

X = smarket.drop('Direction', axis=1)
X = pd.get_dummies(X, drop_first=True)
smarket['Direction'] = np.where(smarket['Direction'] == 'Down', '-1', '1')
y = smarket['Direction']
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```
